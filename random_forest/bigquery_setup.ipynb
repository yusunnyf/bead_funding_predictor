# BigQuery Explorer — broadband-data

```python
from google.cloud import bigquery
import pandas as pd

client = bigquery.Client(project="broadband-data")

datasets = list(client.list_datasets())
print(f"Found {len(datasets)} dataset(s):")
for ds in datasets:
    print(f"  - {ds.dataset_id}")
```

```python
# For each dataset, list tables and preview first 3 rows
for ds in datasets:
    dataset_id = ds.dataset_id
    tables = list(client.list_tables(dataset_id))
    print(f"\n{'='*60}")
    print(f"DATASET: {dataset_id} ({len(tables)} table(s))")
    print(f"{'='*60}")

    for table in tables:
        table_id = table.table_id
        full_id = f"broadband-data.{dataset_id}.{table_id}"
        print(f"\n--- Table: {table_id} ---")
        try:
            df = client.query(f"SELECT * FROM `{full_id}` LIMIT 3").to_dataframe()
            print(f"Columns: {list(df.columns)}")
            display(df)
        except Exception as e:
            print(f"Could not preview: {e}")
```
